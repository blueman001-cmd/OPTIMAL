```python
import matplotlib.pyplot as plt
import numpy as np

# 分类器名称
classifiers = ['PCA', 'IF', 'CBLOF', 'HBOS']

# 三组特征的得分
Raw_Features = [73.76, 74.09, 86.18, 78.82]

NFIDS = [90.83, 90.83, 90.83, 93.22]


# 设置柱状图的宽度和位置
x = np.arange(len(classifiers))
width = 0.2  # 柱子的宽度

# 创建图表
plt.figure(figsize=(8, 5))

# 画三根柱子
plt.bar(x - width/1.5, Raw_Features, width, label='RawFeatures', color='lightcoral')

plt.bar(x + width/1.5, NFIDS, width, label='Our Model', color='lightblue')

# 添加数值标签
for i in range(len(classifiers)):
    plt.text(x[i] - width/1.5, Raw_Features[i] + 0.5, f'{Raw_Features[i]:.2f}', ha='center', va='bottom',fontsize=10)

    plt.text(x[i] + width/1.5, NFIDS[i] + 0.5, f'{NFIDS[i]:.2f}', ha='center', va='bottom',fontsize=10)

# 添加标题和标签
plt.title('NF-CSE-CIC-IDS2018-v2 results (4% contamination)')
# plt.title('NF-UNSW-NB15-v2 results (4% contamination)')
plt.ylabel('Macro F1 (%)')
plt.ylim(0, 100)
plt.yticks([0, 20, 40, 60, 80, 100])

plt.xticks(x, classifiers)

# 添加网格线
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.legend(
    loc='upper center',            # 上方中央基准点
    bbox_to_anchor=(0.5, -0.12),   # 水平居中，向下偏移
    fancybox=True, shadow=False, ncol=3
)

# 保存为图片
plt.savefig("12.jpg", dpi=300, bbox_inches='tight')

# 显示图表
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
methods = ['PCA', 'IF', 'CBLOF', 'HBOS']
nf_unsw_nb15_v2_0 = [85.93, 88.70, 85.94, 92.64]  # Rawfeatures 0%
nf_unsw_nb15_v2_0_model = [70.99, 66.37, 66.48, 71.43]  # Our Model 0%
nf_unsw_nb15_v2_4 = [90.70, 92.69, 90.68, 92.64]  # Rawfeatures 4%
nf_unsw_nb15_v2_4_model = [65.64, 67.03, 71.73, 66.90]  # Our Model 4%
nf_cse_cic_ids2018_v2_0 = [94.43, 94.57, 94.43, 94.47]  # Rawfeatures 0%
nf_cse_cic_ids2018_v2_0_model = [76.84, 81.77, 87.41, 86.28]  # Our Model 0%
nf_cse_cic_ids2018_v2_4 = [90.83, 90.83, 90.83, 93.22]  # Rawfeatures 4%
nf_cse_cic_ids2018_v2_4_model = [73.76, 74.09, 86.18, 78.82]  # Our Model 4%

# Plotting
plt.figure(figsize=(10, 6))

plt.plot(methods, nf_unsw_nb15_v2_0, marker='o', label='NF-UNSW-NB15-v2 0% (OPTIMAL)')
plt.plot(methods, nf_unsw_nb15_v2_0_model, marker='o', label='NF-UNSW-NB15-v2 0% (RAW)')
plt.plot(methods, nf_unsw_nb15_v2_4, marker='o', label='NF-UNSW-NB15-v2 4% (OPTIMAL)')
plt.plot(methods, nf_unsw_nb15_v2_4_model, marker='o',label='NF-UNSW-NB15-v2 4% (RAW)')
plt.plot(methods, nf_cse_cic_ids2018_v2_0, marker='o', label='NF-CSE-CIC-IDS2018-v2 0% (OPTIMAL)')
plt.plot(methods, nf_cse_cic_ids2018_v2_0_model, marker='o', label='NF-CSE-CIC-IDS2018-v2 0% (RAW)')
plt.plot(methods, nf_cse_cic_ids2018_v2_4, marker='o', label='NF-CSE-CIC-IDS2018-v2 4% (OPTIMAL)')
plt.plot(methods, nf_cse_cic_ids2018_v2_4_model, marker='o', label='NF-CSE-CIC-IDS2018-v2 4% (RAW)')

# Adding labels and title
plt.xlabel('Methods')
plt.ylabel('Macro F1 (%)')
plt.title('Macro F1 Scores Across Different Methods and Datasets')
plt.legend()
plt.grid(True)
plt.ylim(0, 100)
plt.savefig("20.jpg", dpi=300, bbox_inches='tight')
# Show plot
plt.show()
```
