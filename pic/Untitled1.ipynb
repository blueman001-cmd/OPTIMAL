```python
import matplotlib.pyplot as plt
import numpy as np

# 数据（Macro F1 %）
datasets = ["UNSW\n(0%)", "UNSW\n(4%)", "CIC\n(0%)", "CIC\n(4%)"]
methods  = ["PCA", "IF", "CBLOF", "HBOS"]

f1_raw = [
    [70.99, 66.37, 68.48, 71.43],  # UNSW 0%
    [65.64, 67.03, 71.73, 66.90],  # UNSW 4%
    [76.84, 81.77, 87.41, 86.28],  # CIC 0%
    [73.76, 74.09, 86.18, 78.82],  # CIC 4%
]

f1_opt = [
    [85.93, 88.70, 85.94, 92.64],
    [90.70, 92.69, 90.68, 92.64],
    [94.43, 94.57, 94.43, 94.47],
    [90.83, 90.83, 90.83, 93.22],
]

# ────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(11, 6.5), dpi=110)

x = np.arange(len(methods))          # 每个方法的基准位置
width = 0.18                         # 每根柱子宽度

for i, ds_name in enumerate(datasets):
    offset = (i - 1.5) * width * 1.15   # 让四个数据集的柱子稍微分开一点
    ax.bar(x + offset, f1_raw[i], width,
           label=f'Raw Features - {ds_name}' if i == 0 else None,
           color='C0', alpha=0.6 + i*0.09, edgecolor='white', linewidth=0.7)
    
    ax.bar(x + offset + width, f1_opt[i], width,
           label=f'OPTIMAL - {ds_name}' if i == 0 else None,
           color='C3', alpha=0.6 + i*0.09, edgecolor='white', linewidth=0.7)

# 美化
ax.set_xlabel('异常检测基线方法', fontsize=12)
ax.set_ylabel('Macro F1 (%)', fontsize=12)
ax.set_title('Raw Features vs OPTIMAL 的 Macro F1 对比\n(四个数据集)', fontsize=13, pad=15, weight='medium')

ax.set_xticks(x + width/2)
ax.set_xticklabels(methods, fontsize=11)

ax.set_ylim(60, 100)
ax.grid(True, axis='y', alpha=0.25, linestyle='--')

# 图例放在底部横向排列
ax.legend(ncol=4, loc='upper center', bbox_to_anchor=(0.5, -0.13),
          fontsize=10, frameon=False, columnspacing=1.2)

plt.tight_layout()
plt.savefig("macro_f1_raw_vs_optimal_one_figure.png", dpi=300, bbox_inches='tight')
plt.show()
```
